## Bayesian approach to finding exoplanets. 

Credit to [Jake Vanderplas](https://github.com/jakevdp/BayesianAstronomy/blob/master/Index.ipynb) for the material from the '*Bayesian Methods in Astronomy workshop at the 227th American Astronomical Society meeting*'.



In the Bayesian approach, observed data is treated as *fixed*, while the model parameters are treated probabilistically. Our degree of knowledge about the Universe, i.e. planets, is 'updated' each time more data is gathered. Formulaically, this can be written as :

$$ \begin{equation}
    P(\text{science}, \text{nuisance parameters}|~\text{data}, \text{background info}).
\end{equation}$$
 
This can be read as: 'what is the probability of this science, along with its nuisance parameters, GIVEN the data and background information'. Science is not done in a vacuum, and generally before a measurement is made, some degree of information about the science is known. 'Nuisance paramters' are parameters we need to fit, but don't generally care about. As an example, if you are finding a planet from radial velocity data, it is important to correctly model the secular motion of the star.

The less wordy version of the formula above can be given as:

$$ \begin{equation}
    P(\bold{\theta, \eta} ~| ~ \bold{\text{I}, {\text{D}}}).
\end{equation}$$

where:

- $\theta$ represents the 'science'; set of parameters we are interested in constraining.
- $\eta$ are the 'nuisance parameters; set of parameters important in the model, but not interesting for scientific result.
- $\text{D}$ represents the 'observed data'.
- $\text{I}$ represents information before observing the data (whatever made you choose the model you're fitting).

This quantity is called the 'posterior probability', but the core problem is there is no way of directly calculating it. We can however calculate $P(\bold{\text{D}} ~| ~ \bold{\theta}, {\text{I}})$, but the two expressions are not equal. This may be calculated through *Bayes' Theorem*.

Bayes' theorem relates the two via:

$$\begin{equation}
P(\theta ~ |~ \text{D}, \text{I}) = \frac{P(D ~|\theta, \text{I})P(\theta ~| \text{I})}{P(\text{D} ~| \text{I})}
\end{equation}$$

where:

- $P(\theta ~|~ \text{D}, \text{I})$ is the **posterior**. This is the quantity we want to compute: our knowledge of the model given the data & background knowledge (including the choice of model).

- $P(\text{D} ~|~ \theta, \text{I})$ is the **likelihood**. This measures the probability of seeing our data given the model. This is identical to the quantity maximized in frequentist maximum-likelihood approaches.

- $P(\theta ~|~ \text{I})$ is the **prior**. This encodes any knowledge we had about the answer before measuring the current data.

- $P(\text{D} ~|~ \text{I})$ is the **Fully Marginalized Likelihood**. In the context of model fitting, it acts as a normalization constant and in most cases can be ignored.
